## Passo 1 — Importar bibliotecas

O K-Means organiza observações segundo a proximidade entre seus indicadores.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

## Passo 2 — Criar municípios fictícios

Os 36 municípios ilustrativos terão indicadores diferentes de envelhecimento e renda.

```python
rng = np.random.default_rng(10)
n = 36

dados = pd.DataFrame({
    "municipio": [f"Município {i:02d}" for i in range(1, n + 1)],
    "pct_idosos": rng.uniform(8, 30, n),
    "renda_media": rng.uniform(1100, 3400, n)
})
print(dados.head())
```

## Passo 3 — Preparar as variáveis

Como renda e percentual têm escalas distintas, padronizamos antes do agrupamento.

```python
colunas = ["pct_idosos", "renda_media"]
padronizador = StandardScaler()
X = padronizador.fit_transform(dados[colunas])

modelo = KMeans(n_clusters=3, random_state=42, n_init=10)
dados["grupo"] = modelo.fit_predict(X)

print(dados.groupby("grupo")[colunas].mean().round(1))
```

## Passo 4 — Visualizar os perfis

Cada ponto representa um município. A cor indica o grupo encontrado pelo algoritmo.

```python
plt.figure(figsize=(8, 5))
for grupo, tabela in dados.groupby("grupo"):
    plt.scatter(
        tabela["pct_idosos"],
        tabela["renda_media"],
        label=f"Grupo {grupo}",
        s=55
    )

plt.xlabel("Pessoas com 60 anos ou mais (%)")
plt.ylabel("Renda média fictícia (R$)")
plt.title("Perfis municipais — dados sintéticos")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
```

## Passo 5 — Experimentar outro número de grupos

Os agrupamentos mudam quando definimos outra quantidade de grupos.

```python
for quantidade in [2, 3, 4]:
    modelo = KMeans(
        n_clusters=quantidade, random_state=42, n_init=10
    )
    grupos = modelo.fit_predict(X)
    print(f"{quantidade} grupos:", np.bincount(grupos))
```
