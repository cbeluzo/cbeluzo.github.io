## Passo 1 — Preparar o ambiente

Vamos usar apenas bibliotecas comuns de análise de dados.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
print("Ambiente pronto!")
```

## Passo 2 — Criar dados sintéticos

Os valores são fictícios e servem apenas à exploração da análise.

**O que observar:** O que é um DataFrame? É uma tabela com linhas (municípios) e colunas (indicadores).

```python
rng = np.random.default_rng(42)
municipios = ["Município A", "Município B", "Município C", "Município D", "Município E"]
pop = np.array([10000, 20000, 12000, 18000, 15000])
pct_idosos = np.array([10, 18, 14, 23, 8])
df = pd.DataFrame({"municipio": municipios, "populacao": pop, "pct_60mais": pct_idosos})
df["pessoas_60mais"] = (df.populacao * df.pct_60mais / 100).round().astype(int)
df
```

## Passo 3 — Calcular indicadores

A média simples dos percentuais municipais não equivale à proporção populacional agregada.

**O que observar:** O indicador agregado usa o total de pessoas de 60 anos ou mais dividido pela população total. Compare com a média aritmética dos percentuais municipais.

```python
total = df["populacao"].sum()
pessoas_idosas = df["pessoas_60mais"].sum()
print("População total:", total)
print("Proporção ponderada de pessoas 60+ (%):", round(100*pessoas_idosas/total, 2))
```

## Passo 4 — Visualizar

Compare as barras: o gráfico não permite concluir por que os municípios diferem.

**O que observar:** A altura de cada barra representa um percentual municipal; observe também as unidades dos eixos.

```python
ax = df.plot.bar(x="municipio", y="pct_60mais", legend=False, figsize=(9, 4))
ax.set_ylabel("População de 60 anos ou mais (%)")
ax.set_xlabel("Município (fictício)")
plt.xticks(rotation=25, ha="right")
plt.tight_layout()
plt.show()
```
